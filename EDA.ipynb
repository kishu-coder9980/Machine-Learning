```python
sns.countplot(data=df,x="species")
plt.title("Number of Flowers by Species")
plt.xlabel("Species")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="sepal_length",
    bins=10,
    kde=True
)
```

```python
sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)
plt.title("Petal Length by Species")
plt.show()
```
